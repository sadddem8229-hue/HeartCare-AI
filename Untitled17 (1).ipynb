```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

PROJECT_NAME = "Heart Disease Prediction"
DATA_URL = "https://raw.githubusercontent.com/sharmaroshan/Heart-UCI-Dataset/master/heart.csv"

df = pd.read_csv(DATA_URL, encoding="utf-8-sig")

print("Project:", PROJECT_NAME)
print("Data Shape:", df.shape)
print(df.head())
print("Lab 1 - Code 1 completed!")
```

```python
sns.set_theme(style="whitegrid")

print("Missing values:", int(df.isna().sum().sum()))
print("Duplicate rows:", int(df.duplicated().sum()))
df = df.drop_duplicates().reset_index(drop=True)

counts = df["target"].value_counts().sort_index()
print("No disease (0):", counts[0], "| Disease (1):", counts[1])

fig, ax = plt.subplots(figsize=(5, 3.8))
bars = ax.bar(["No disease", "Disease"], counts.values,
              color=["#4C9F9A", "#E4572E"], width=0.55)
ax.bar_label(bars, fontweight="bold")
ax.set_title("Class balance", fontweight="bold")
plt.show()
print("Lab 1 - Code 2 completed!")
```

```python
cols = ["age", "thalach", "trestbps", "chol", "oldpeak"]
fig, axes = plt.subplots(1, 5, figsize=(18, 3.6))
for ax, c in zip(axes, cols):
    for t, color, lab in [(0, "#4C9F9A", "No disease"), (1, "#E4572E", "Disease")]:
        sns.kdeplot(df.loc[df.target == t, c], ax=ax, fill=True,
                    alpha=0.45, color=color, label=lab)
    ax.set_title(c, fontweight="bold"); ax.set_xlabel(""); ax.set_ylabel("")
axes[0].legend()
plt.show()

plt.figure(figsize=(9, 7))
sns.heatmap(df.corr(), cmap="coolwarm", center=0, annot=True, fmt=".2f",
            annot_kws={"size": 8}, linewidths=1, linecolor="white")
plt.title("Correlation matrix", fontweight="bold")
plt.show()
print("Lab 1 - Code 3 completed!")
```

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=["target"])
y = df["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))
print("Lab 1 - Code 4 completed!")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

def make_pipeline(model):
    return Pipeline([
        ("impute", SimpleImputer(strategy="median")),
        ("scale", StandardScaler()),
        ("model", model),
    ])

models = {
    "Logistic Regression": make_pipeline(LogisticRegression(max_iter=1000, random_state=42)),
    "Random Forest": make_pipeline(RandomForestClassifier(n_estimators=300, random_state=42)),
}

for name, m in models.items():
    m.fit(X_train, y_train)
    print("Trained:", name)
print("Lab 1 - Code 5 completed!")
```
